## Difference Check

```python
def get_difference(obj, col1, col2):
	return obj.groupby("carrier").apply(lambda x: (x[col1] - x[col2]).min())
```

```python
get_difference(n41.generators, 'p_nom_opt', 'p_nom')
```

```python
get_difference(n0.generators, 'p_nom_opt', 'p_nom')
```
